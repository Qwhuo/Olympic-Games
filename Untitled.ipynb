```python
# участвовало ли меньше человек в 1980 году из-за бойкотов?
def count_expect_part(year, prev=None, nxt=None, data=p_by_year):
    if prev is None:
        prev = year - 4
    if nxt is None:
        nxt = year + 4
    N_exp = (p_by_year.loc[p_by_year.index == prev, "total"].iloc[0] + p_by_year.loc[p_by_year.index == nxt, "total"].iloc[0]) / 2
    N_act = p_by_year.loc[p_by_year.index == year, "total"].iloc[0]
    print(f">> Expected amount of participants in {year}: {N_exp}\n>> Actual: {N_act}\n>> Diff: {(N_exp - N_act) / N_exp * 100:.2f}%")

count_expect_part(1980)
```
